# Canvas Setup Test Notebook

This notebook demonstrates the `setup_canvas()` function from `rua.canvas` module, which creates clean diagram figures with no axes or padding.

## 1. Import Required Libraries

```python
import matplotlib.pyplot as plt
from rua import setup_canvas
from rua.components.stellar import star, zams, wolf_rayet
from rua.components.binary import binary, HMS_HMS, common_envelope
from rua.components.compact import NS, BH, WD
```

## 2. Basic Canvas Setup with Single Star

Create a clean canvas with default settings and draw a simple star.

```python
fig, ax = setup_canvas(height=3, width=3)

# Draw a star in the center
star(1.5, 1.5, size=0.8, color='yellow', label_text='Main Sequence Star', ax=ax)

plt.show()
```

## 3. Canvas with Binary System

Create a larger canvas and draw a binary star system.

```python
fig, ax = setup_canvas(width=12, height=6, xlim=(-6, 6), ylim=(-3, 3))

# Draw a high-mass binary system
HMS_HMS(0, 0, size_primary=0.6, size_secondary=0.5, 
        separation=2.0, label_text='HMS Binary', ax=ax)

plt.show()
```

## 4. Canvas with Multiple Stellar Objects

Demonstrate different stellar phases on one canvas.

```python
fig, ax = setup_canvas(width=14, height=6, xlim=(-7, 7), ylim=(-3, 3))

# Draw different stellar phases
zams(-4, 0, size=0.5, label_text='ZAMS', ax=ax)
wolf_rayet(0, 0, size=0.6, label_text='Wolf-Rayet', ax=ax)
BH(4, 0, size=0.4, label_text='Black Hole', ax=ax)

plt.show()
```

## 5. Canvas with Common Envelope Phase

Show a special binary evolution phase with custom canvas settings.

```python
fig, ax = setup_canvas(width=10, height=8, xlim=(-4, 4), ylim=(-4, 4))

# Draw a common envelope phase
common_envelope(0, 0, primary=star, secondary=star,
                envelope_size=1.5, label_text='Common Envelope', ax=ax)

plt.show()
```

## 6. Comparison: With vs Without Axes

Compare the clean canvas (default) vs. a canvas with axes visible.

```python
# Canvas with axes removed (default behavior)
fig1, ax1 = setup_canvas(width=8, height=5, remove_axes=True)
star(4, 4, size=0.7, color='yellow', label_text='Clean Canvas', ax=ax1)
plt.title('With Axes Removed (Clean)')
plt.show()

# Canvas with axes visible
fig2, ax2 = setup_canvas(width=8, height=5, remove_axes=False)
star(3, 3, size=0.7, color='yellow', label_text='With Axes', ax=ax2)
plt.title('With Axes Visible')
plt.show()
```

## Summary

The `setup_canvas()` function provides:
- Clean, professional-looking diagrams with no padding or axes
- Customizable figure size and axis limits
- Equal aspect ratio by default (can be changed)
- Simple interface that returns both `fig` and `ax` objects

Perfect for creating publication-ready stellar evolution diagrams!
